```python
import tensorflow as tf
from tensorflow.keras import layers
import numpy as np

# Define input shape: sequence length = 3, features = 1
inputs = tf.keras.Input(shape=(3, 1))

# LSTM with return_sequences=True returns output at each timestep
lstm_1 = layers.LSTM(1, return_sequences=True)(inputs)
model_lstm_1 = tf.keras.models.Model(inputs=inputs, outputs=lstm_1)

# LSTM with return_sequences=False returns output only at last timestep
inputs = tf.keras.Input(shape=(3, 1))
lstm_2 = layers.LSTM(1, return_sequences=False)(inputs)
model_lstm_2 = tf.keras.models.Model(inputs=inputs, outputs=lstm_2)

# Sample input data: one sequence of length 3 with 1 feature each
data = np.array([[[0.1],
                  [0.2],
                  [0.3]]])

print("Input data:")
print(data)

print("\nOutput when return_sequences=True:")
print(model_lstm_1.predict(data))

print("\nOutput when return_sequences=False:")
print(model_lstm_2.predict(data))
```

```python
model = tf.keras.Sequential()
model.add(layers.Dense(10, activation='softmax'))

# Compile the model with optimizer, loss function, and metrics
model.compile(
    optimizer=tf.keras.optimizers.Adam(0.001),
    loss=tf.keras.losses.categorical_crossentropy,   # fixed typo here
    metrics=[tf.keras.metrics.categorical_accuracy]
)
```

```python
import numpy as up
train_x = np.random.random((1000, 36))
train_y = np.random.random((1000, 10))
val_x = np.random.random((200, 36))
val_y = np.random.random((200, 10))
model.fit(train_x, train_y, epochs=10, batch_size=100,
validation_data=(val_x, val_y))
```

```python
dataset = tf.data.Dataset.from_tensor_slices((train_x, train_y))
dataset = dataset.batch(32)
dataset = dataset.repeat()
val_dataset = tf.data.Dataset.from_tensor_slices((val_x, val_y))
val_dataset = val_dataset.batch(32)
val_dataset = val_dataset.repeat()
model.fit(dataset, epochs=10, steps_per_epoch=30,
validation_data=val_dataset, validation_steps=3)
```

```python
import tensorflow as tf

# Set hyperparameters
Epochs = 10

# Define a function for dynamically setting the learning rate
def lr_scheduler(epoch):
    if epoch > 0.9 * Epochs:
        lr = 0.0001
    elif epoch > 0.5 * Epochs:
        lr = 0.001
    elif epoch > 0.25 * Epochs:
        lr = 0.01
    else:
        lr = 0.1
    print(f"Epoch {epoch+1}: Learning rate = {lr}")
    return lr

callbacks = [
    # Early stopping to stop training if validation loss doesn't improve
    tf.keras.callbacks.EarlyStopping(
        monitor='val_loss',
        min_delta=1e-2,
        patience=2
    ),
    # Save the best model during training
    tf.keras.callbacks.ModelCheckpoint(
        filepath='testmodel_{epoch}.h5',  # fixed string formatting braces
        save_best_only=True,
        monitor='val_loss'
    ),
    # Dynamically change the learning rate
    tf.keras.callbacks.LearningRateScheduler(lr_scheduler),
    # Use TensorBoard for visualization
    tf.keras.callbacks.TensorBoard(log_dir='./logs')  # fixed missing quotes
]

# Assuming train_x, train_y, val_x, val_y are already defined datasets
model.fit(
    train_x, train_y,
    batch_size=16,
    epochs=Epochs,
    callbacks=callbacks,
    validation_data=(val_x, val_y)
)
```

```python
#Model evaluation
test_x = np.random.random((1000, 36))
test_y = np.random.random((1000, 10))
model.evaluate(test_x, test_y, batch_size=32)
```
